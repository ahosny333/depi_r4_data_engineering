```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.options import Options
import time
import pandas as pd

list_jobs = ['محلل بيانات','محاسب','مدخل بيانات']
job = list_jobs[2]

options = Options()
options.add_argument("--disable-blink-features=AutomationControlled")
options.add_experimental_option("excludeSwitches", ["enable-automation"])
options.add_experimental_option('useAutomationExtension', False)

driver = webdriver.Chrome(options=options)
driver.maximize_window()

driver.get('https://forasna.com/')
time.sleep(2)

driver.find_element(By.NAME, "query").send_keys(job)
driver.find_element(By.XPATH, "//span[@class='jobs-text']").click()
time.sleep(2)

job_results = []

for page in range(3):
    jobs = driver.find_elements(By.CLASS_NAME, "result-wrp")
    for job in jobs:
        try:
            title = job.find_element(By.CLASS_NAME, "job-title").text
            company = job.find_element(By.CLASS_NAME, "company-name").text
            location = job.find_element(By.CLASS_NAME, "location").text
            description = job.find_element(By.CLASS_NAME, "job-details__title").text

            job_results.append({
                'title': title,
                'company': company,
                'location': location,
                'description': description
            })
        except:
            continue

    try:
        driver.find_element(By.XPATH, "//ul[@class='pagination desktop-pagination hidden-xs']//a[contains(text(),'التالى')]").click()
        time.sleep(2)
    except:
        break

driver.quit()

df = pd.DataFrame(job_results)
df.to_csv('web.csv', index=False)
```

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.options import Options
import time
import pandas as pd

list_jobs = ['data analysis','data entry','data engineering']
job = list_jobs[2]

options = Options()
options.add_argument("--disable-blink-features=AutomationControlled")
options.add_experimental_option("excludeSwitches", ["enable-automation"])
options.add_experimental_option('useAutomationExtension', False)

driver = webdriver.Chrome(options=options)
driver.maximize_window()

driver.get('https://www.arabjobs.com/en')
time.sleep(2)

driver.find_element(By.ID, "txtKeyword").send_keys(job)
driver.find_element(By.XPATH, "//span[@class='fa fa-search']").click()
time.sleep(2)

job_results = []

for page in range(3):
    jobs = driver.find_elements(By.CLASS_NAME, "block")
    for job in jobs:
        
            title = job.find_element(By.CLASS_NAME, "font-blocks").text
            company = job.find_element(By.CLASS_NAME, "col-xs-12").text
            category = job.find_element(By.CLASS_NAME, "category").text
            description = job.find_element(By.CLASS_NAME, "category").text

            job_results.append({
                'title': title,
                'company': company,
                'category': category,
                'description': description
            })
        
            

    try:
        driver.find_element(By.XPATH, "li:nth-child(9) a:nth-child(1)").click()
        time.sleep(2)
    except:
        break

driver.quit()

df = pd.DataFrame(job_results)
df.to_csv('webbbb.csv', index=False)
```
