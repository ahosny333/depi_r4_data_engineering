Adzuna

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.service import Service
from webdriver_manager.chrome import ChromeDriverManager
import pandas as pd
import time

from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC

job_queries = ["Accountant", "Data Engineer", "Data Analyst"]

driver = webdriver.Chrome(service=Service(ChromeDriverManager().install()))
driver.set_page_load_timeout(60)

wait = WebDriverWait(driver, 20)

all_jobs = []

for job_query in job_queries:

    print(f"\n Searching for: {job_query}")

    search_query = job_query.replace(" ", "+")
    url = f"https://www.adzuna.com/search?q={search_query}"

    driver.get(url)

    wait.until(
        EC.presence_of_element_located((By.CSS_SELECTOR, "h2"))
    )

    time.sleep(3)

    cards = driver.find_elements(By.CSS_SELECTOR, "div")

    for card in cards:

        try:
            title_element = card.find_element(By.CSS_SELECTOR, "h2 a")
            title = title_element.text
            link = title_element.get_attribute("href")
        except:
            continue

        try:
            company = card.find_element(By.CLASS_NAME, "ui-company").text
        except:
            company = ""

        try:
            location = card.find_element(By.CLASS_NAME, "ui-location").text
        except:
            location = ""

        try:
            salary = card.find_element(By.CLASS_NAME, "ui-salary").text
        except:
            salary = ""

        all_jobs.append([
            job_query,   
            title,
            company,
            location,
            salary,
            link
        ])

driver.quit()

df = pd.DataFrame(all_jobs, columns=[
    "search_keyword",
    "title",
    "company",
    "location",
    "salary",
    "link"
])

print(df)

df.to_csv("all_jobs.csv", index=False)

print("\n Saved to all_jobs.csv")
```

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
import pandas as pd
import time

keyword = "accountant"

driver = webdriver.Chrome()

driver.get(f"https://www.bayt.com/en/egypt/jobs/{keyword}-jobs/")

time.sleep(5)

def scrape_bayt():

    jobs = []

    job_cards = driver.find_elements(By.CSS_SELECTOR, "li[data-js-job]")

    print("Jobs found:", len(job_cards))

    for job in job_cards[:20]:

        try:
            title = job.find_element(By.CSS_SELECTOR, ".t-default, .t-large").text
        except:
            title = "N/A"

        try:
            company = job.find_element(By.CSS_SELECTOR, ".t-bold").text
        except:
            company = "N/A"

        try:
            location = job.find_elements(By.CLASS_NAME, "t-mute")[1].text
        except:
            location = "N/A"

        try:
            link = job.find_element(By.TAG_NAME, "a").get_attribute("href")
        except:
            link = "N/A"

        try:
            salary = job.find_element(By.CSS_SELECTOR, ".jb-label-salary").text
        except:
            salary = "N/A"

        try:
            career = job.find_element(By.CSS_SELECTOR, ".jb-label-careerlevel").text
        except:
            career = "N/A"


        jobs.append({
            "Job Title": title,
            "Company": company,
            "Location": location,
            "Salary": salary,
            "Career Level": career,
            "Apply Link": link
        })

    return jobs


data = scrape_bayt()

df = pd.DataFrame(data)

print("\nJobs Found:", len(df))
print(df)

driver.quit()
```
