```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
import time
import pandas as pd

list_jobs = ['محلل بيانات', 'محاسب', 'مدخل بيانات']
job = list_jobs[2]

options = Options()
options.add_argument("--disable-blink-features=AutomationControlled")
options.add_experimental_option("excludeSwitches", ["enable-automation"])
options.add_experimental_option('useAutomationExtension', False)

driver = webdriver.Chrome(options=options)
driver.maximize_window()

driver.get("https://forasna.com/%D9%88%D8%B8%D8%A7%D8%A6%D9%81-%D8%AE%D8%A7%D9%84%D9%8A%D8%A9?query=")

wait = WebDriverWait(driver, 10)

try:
    search_input = wait.until(EC.element_to_be_clickable((By.CLASS_NAME, "tt-input")))
    search_input.click()
    time.sleep(0.5)
    search_input.send_keys(job)
    search_input.send_keys(Keys.RETURN)

except Exception:
    try:
        inputs = driver.find_elements(By.CLASS_NAME, "tt-input")
        search_input = next(el for el in inputs if el.is_displayed())
        search_input.click()
        search_input.send_keys(job)
        search_input.send_keys(Keys.RETURN)

    except Exception:
        search_input = driver.find_element(By.CLASS_NAME, "tt-input")
        driver.execute_script("arguments[0].value = arguments[1];", search_input, job)
        driver.execute_script("arguments[0].dispatchEvent(new Event('input'));", search_input)
        driver.find_element(By.CSS_SELECTOR, ".input-group-btn button").click()

time.sleep(2)

job_results = []

for page in range(3):

    try:
        wait.until(EC.presence_of_element_located((By.CLASS_NAME, "result-wrp")))
    except:
        break

    job_cards = driver.find_elements(By.CLASS_NAME, "result-wrp")

    for job_card in job_cards:
        try:
            title    = job_card.find_element(By.CLASS_NAME, "job-title").text
            company  = job_card.find_element(By.CLASS_NAME, "company-name").text
            location = job_card.find_element(By.CLASS_NAME, "location").text

            link = job_card.find_element(By.TAG_NAME, "a").get_attribute("href")

            try:
                experience = job_card.find_element(By.CSS_SELECTOR,"#search-results-container > div.content-card.card-has-jobs > div:nth-child(3) > div.col-md-10.col-sm-12.col-xs-12.item-details > div.mobile-card-body > div.job-details > span:nth-child(1)").text
            except:
                experience = None

            details = job_card.find_elements(By.CLASS_NAME, "job-details")
            salary = None
            if len(details) > 1:
                salary_text = details[1].text
                if 'الراتب غير معلن' not in salary_text:
                    salary = salary_text[15:].strip()

            print(f"title: {title}")
            print(f"salary: {salary}")
            print(f"experience: {experience}")
            print('-' * 50)

            job_results.append({
                'title':      title,
                'company':    company,
                'location':   location,
                'link':       link,
                'experience': experience,
                'salary':     salary
            })

        except Exception as e:
            print(f"error: {e}")
            continue   

driver.quit()

df = pd.DataFrame(job_results)
df.to_csv('forsna.csv', index=False, encoding='utf-8-sig')
```
